```python
def codebook(fichier, noms):
    """Code, nom explicite, libellé Stata et modalités — lit seulement l'en-tête du .dta."""
    r = pd.io.stata.StataReader(path / fichier)
    libelles = r.variable_labels()
    valeurs = r.value_labels()
    jeux = dict(zip(r._varlist, r._lbllist))
    return pd.DataFrame({
        "variable": r._varlist,
        "nom_explicite": [noms.get(v, "") for v in r._varlist],
        "libelle": [libelles.get(v, "") for v in r._varlist],
        "modalites": [valeurs.get(jeux.get(v)) for v in r._varlist],
    })


for fichier, noms in [("Base_Individus.dta", INDIVIDUS), ("Base_Menage.dta", MENAGE),
                      ("Base_avoirs_du_menage.dta", AVOIRS)]:
    cb = codebook(fichier, noms)
    cb.to_csv(f"codebook_{fichier.replace('.dta', '')}.csv", index=False)
    print(fichier, len(cb), "variables,", (cb.nom_explicite != "").sum(), "documentées")

cb.head(10)
```

## Dataset de travail IPM

Sélection des seules variables utiles au calcul de l'IPM, dans les 4 bases, puis fusion.

Deux partis pris :
- **rechargement en codes bruts** (`convert_categoricals=False`) : les règles de privation se
  écrivent sur des codes (`type_sanitaire == 1`), pas sur des libellés. Les `Base_*` chargées
  plus haut restent en libellés pour l'exploration ;
- **niveau individu** (64 491 lignes) : les caractéristiques du ménage sont recopiées sur chaque
  membre. Aucune agrégation n'est faite ici, car passer au ménage suppose des conventions
  (« au moins un membre privé », qui est éligible, que faire des manquants) qui se décident
  indicateur par indicateur.

Les avoirs (format long, 45 biens × ménage) sont mis à plat sur les 8 biens du MPI.

```python
CLE = ["grappe", "menage", "vague"]

# --- variables utiles à l'IPM -------------------------------------------------
IPM_INDIVIDUS = [
    "id_membre", "sexe", "lien_parente_cm", "annee_naissance", "mois_naissance",
    "age_declare", "present_menage", "reside_6_mois",
    "acte_naissance",                                            # identification
    "lit_francais", "ecrit_francais", "a_frequente_ecole",       # éducation
    "scolarise_2020_2021", "scolarise_2021_2022", "niveau_instruction",
    "diplome_plus_eleve",
    "probleme_sante_30j", "consultation_sante_30j",              # santé
    "raison_non_consultation", "assurance_maladie", "moustiquaire",
    "travail_champ_7j", "travail_commerce_7j", "travail_salarie_7j",   # emploi
    "travail_apprenti_7j", "a_travaille_7j", "emploi_mais_absent_7j",
    "raison_non_travail_7j", "recherche_emploi_30j_a", "recherche_emploi_30j_b",
    "disponible_emploi", "delai_disponibilite",
]
IPM_MENAGE = [
    "id_menage", "ponderation_menage", "taille_menage", "region", "milieu",
    "depense_par_tete", "seuil_pauvrete_national",               # contrôle monétaire
    "nb_pieces", "materiau_mur", "materiau_toit", "materiau_sol",
    "source_eau_boisson_seche", "distance_source_eau_seche",
    "source_eclairage", "combustible_gaz", "combustible_electricite",
    "type_sanitaire", "sanitaire_partage",
]
IPM_FIES = [v for v in SECURITE_ALIMENTAIRE.values() if v.startswith("fies")]
BIENS_IPM = {16: "refrigerateur", 19: "radio", 20: "television", 28: "voiture",
             29: "moto", 30: "bicyclette", 35: "telephone_portable", 37: "ordinateur"}


def charger_brut(fichier, garder):
    """Recharge le .dta en codes bruts, limité aux colonnes voulues, et renomme."""
    noms = BASES[fichier]
    codes = {v: k for k, v in noms.items()}
    d = pd.read_stata(path / fichier, columns=[codes[c] for c in garder],
                      convert_categoricals=False)
    return d.rename(columns=noms)


individus = charger_brut("Base_Individus.dta", CLE + IPM_INDIVIDUS)
menages = charger_brut("Base_Menage.dta", CLE + IPM_MENAGE)
fies = charger_brut("Base_securite_alimentaire.dta", CLE + IPM_FIES)
fies = fies.dropna(subset=CLE)          # 728 lignes entièrement vides dans cette base

# avoirs : format long (45 biens x ménage) -> une colonne par bien du MPI
avoirs = charger_brut("Base_avoirs_du_menage.dta", CLE + ["code_bien", "possede_bien"])
biens = (avoirs[avoirs.code_bien.isin(BIENS_IPM)]
         .assign(possede=lambda d: (d.possede_bien == 1).astype(int),
                 bien=lambda d: "possede_" + d.code_bien.map(BIENS_IPM))
         .pivot_table(index=CLE, columns="bien", values="possede", aggfunc="max")
         .reset_index())

ipm_data = (individus
            .merge(menages, on=CLE, how="left", validate="m:1", indicator="_menage")
            .merge(biens, on=CLE, how="left", validate="m:1")
            .merge(fies, on=CLE, how="left", validate="m:1", indicator="_fies"))

# contrôles d'appariement : tout doit être apparié
print("individus :", len(individus), "| ménages :", len(menages),
      "| ménages dans les avoirs :", len(biens), "| ménages FIES :", len(fies))
print("appariement ménage :", ipm_data._menage.value_counts().to_dict())
print("appariement FIES   :", ipm_data._fies.value_counts().to_dict())

ipm_data = ipm_data.drop(columns=["_menage", "_fies"])
print("\ndataset final :", ipm_data.shape)
ipm_data.head()
```

```python
ipm_data.to_stata("ipm_data_ehcvm2021.dta", write_index=False, version=118)
print("enregistré : ipm_data_ehcvm2021.dta")

# taux de non-réponse, pour repérer les variables à conventions (filtres du questionnaire)
ipm_data.isna().mean().sort_values(ascending=False).head(15).round(3)
```
